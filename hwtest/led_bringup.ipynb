````python
import logging, time
logging.basicConfig(level=logging.INFO, format="  [%(levelname)s] %(message)s", force=True)
from syncboard import SyncBoard, FeedbackMode

CH = 5
sb = SyncBoard.connect()
sb.ping()
````

**Stop** (run any time)

````python
sb.leds.all_off(); sb.disable()
````

Attach, stay disabled. HB low, rails off, LED dark.

````python
sb.disable()
sb.attach_led_board(True)
sb.status()
````

````python
sb.enable()
sb.leds.set_level(CH, 0.1)
sb.leds.on(CH)
time.sleep(0.002)
sb.leds.all_off();
sb.disable()
````

````python
sb.leds.pulse(CH, 1000)
````

````python
sb.enable()
MAX_A = 0.3
sb.leds.calibrate(CH, MAX_A)
sb.leds.off(CH)
sb.disable()
````
